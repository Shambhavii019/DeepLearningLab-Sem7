## Generating Dummy Training Data

To demonstrate the model's `fit()` method, we need to create some dummy training data that matches the input and output shapes of our defined model.

````python
import numpy as np
import tensorflow as tf

# Define the number of samples for dummy data
num_samples = 100

# Generate dummy input features (X_train) with 4 features
X_train = np.random.rand(num_samples, 4).astype(np.float32);

# Generate dummy target labels (y_train) for 3 classes (0, 1, or 2)
# Then, one-hot encode these labels
y_train_labels = np.random.randint(0, 3, num_samples)
y_train = tf.keras.utils.to_categorical(y_train_labels, num_classes=3)

print("Shape of X_train:", X_train.shape)
print("Shape of y_train:", y_train.shape)

print("\nFirst 5 rows of X_train:\n", X_train[:5])
print("\nFirst 5 rows of y_train (one-hot encoded):\n", y_train[:5])
````

## Model Training with Dummy Data

Now, let's train the model using the generated `X_train` and `y_train`.

````python
from tensorflow import keras
from tensorflow.keras import layers
import tensorflow as tf # Ensure tf is imported for keras utilities, though not strictly needed for model building here.

# Model definition (copied from the intended model creation cell)
model = keras.Sequential([
    layers.Input(shape=(4,)),
    layers.Dense(16, activation='relu'),
    layers.Dense(8, activation='relu'),
    layers.Dense(3, activation='softmax')
])

# Model compilation (copied from the intended model compilation cell)
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(X_train, y_train,
                    epochs=10,
                    batch_size=32,
                    verbose=1)

print("\nModel training complete.")
````
